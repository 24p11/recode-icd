 inspect_code — mode d'emploi

`inspect_code(codes, ctx=None)` affiche un rapport texte en 4 blocs
pour un ou plusieurs codes CIM-10 :

- **BLOC 1 — Identité** : libellés OFS/ANS, position hiérarchique
  (chapitre / bloc / catégorie), type.
- **BLOC 2 — Sources brutes** : ce que dit chaque source avant fusion
  (OFS, ANS, ORPHANET avec relation E/NTBT, Index CIM-10 vol3, AP-HP
  par spécialité).
- **BLOC 3 — Dague/astérisque** : les associations de la table DAGSTAR
  enrichie (partenaire, redundancy_level, niveaux présents).
- **BLOC 4 — Résultat final** : les lignes du CSV maître pour ce code,
  réparties par (type, source), avec source_level / inherited_from_code.

**Résolution des codes** :
- `"A18.1"` → le code exact (s'il existe dans le CSV).
- `"A18"` → préfixe : tous les codes du CSV commençant par `A18`.
- `["A18.1", "N33.0"]` → liste de codes/préfixes.

`ctx=None` recharge automatiquement le contexte avec les sources
externes. Ici on passe `ctx=ctx` pour réutiliser le contexte déjà
chargé (plus rapide).

# Exploration du contenu de la table DAGSTAR et du recouvrement avec les autres sources d'informations.

```python
import polars as pl

# =====================================================================
# Investigation 1 — Couverture des LID DAGSTAR par les autres tables
# =====================================================================
print("=" * 70)
print("INVESTIGATION 1 — Couverture des LID DAGSTAR")
print("=" * 70)

dagstar = ctx.ofs["dagstar"]
total_dagstar = dagstar.height
total_lid_uniques = dagstar.select("LID").n_unique()
print(f"\nDAGSTAR : {total_dagstar} lignes, {total_lid_uniques} LID uniques")

# LID uniques de DAGSTAR
lid_dagstar = set(dagstar.select("LID").to_series().to_list())

# Couverture par LIBELLE
lid_libelle = set(ctx.ofs["libelle"].select("LID").to_series().to_list())
inter_libelle = lid_dagstar & lid_libelle
print(f"\nLID DAGSTAR présents dans LIBELLE : {len(inter_libelle)} / {len(lid_dagstar)}")
print(f"  (donc {len(lid_dagstar) - len(inter_libelle)} LID DAGSTAR absents de LIBELLE)")

# Couverture par INCLUDE
if "include" in ctx.ofs:
    lid_include = set(ctx.ofs["include"].select("LID").to_series().to_list())
    inter_include = lid_dagstar & lid_include
    print(f"\nLID DAGSTAR présents dans INCLUDE : {len(inter_include)} / {len(lid_dagstar)}")

# Couverture par EXCLUDE
if "exclude" in ctx.ofs:
    lid_exclude = set(ctx.ofs["exclude"].select("LID").to_series().to_list())
    inter_exclude = lid_dagstar & lid_exclude
    print(f"\nLID DAGSTAR présents dans EXCLUDE : {len(inter_exclude)} / {len(lid_dagstar)}")

# Couverture par DESCR
if "descr" in ctx.ofs:
    lid_descr = set(ctx.ofs["descr"].select("LID").to_series().to_list())
    inter_descr = lid_dagstar & lid_descr
    print(f"\nLID DAGSTAR présents dans DESCR : {len(inter_descr)} / {len(lid_dagstar)}")

# LID DAGSTAR couverts par AU MOINS une des tables ci-dessus
unions = lid_libelle
if "include" in ctx.ofs:
    unions = unions | lid_include
if "exclude" in ctx.ofs:
    unions = unions | lid_exclude
if "descr" in ctx.ofs:
    unions = unions | lid_descr
couverts = lid_dagstar & unions
non_couverts = lid_dagstar - unions
print(f"\nLID DAGSTAR couverts par au moins UNE des tables : {len(couverts)} / {len(lid_dagstar)}")
print(f"LID DAGSTAR non couverts (orphans) : {len(non_couverts)}")
if non_couverts:
    print(f"  Exemples : {list(non_couverts)[:5]}")

# =====================================================================
# Investigation 2 — Distribution des LID DAGSTAR par source LIBELLE
# =====================================================================
print("\n" + "=" * 70)
print("INVESTIGATION 2 — Source LIBELLE des LID DAGSTAR")
print("=" * 70)

# Jointure DAGSTAR → LIBELLE pour récupérer le source
dagstar_with_source = dagstar.join(
    ctx.ofs["libelle"].select(["LID", "source"]),
    on="LID",
    how="left",
)
distrib_source = (
    dagstar_with_source
    .group_by("source")
    .agg(pl.len().alias("n"))
    .sort("n", descending=True)
)
print("\nDistribution du source LIBELLE pour les LID de DAGSTAR :")
print(distrib_source)

# Légende : S=systematique, D=descripteur, I=inclusion, E=exclusion, N=note
print("\nLégende habituelle OFS : S=systematique, D=descripteur, I=inclusion, E=exclusion, N=note")

# =====================================================================
# Investigation 3 — Position du LID : côté SID (descripteur) vs côté assoc
# =====================================================================
print("\n" + "=" * 70)
print("INVESTIGATION 3 — Position du LID dans la mécanique dague/astérisque")
print("=" * 70)

# Pour chaque ligne DAGSTAR, le LID est rattaché à un SID (via LIBELLE).
# Vérifier : le LID est-il rattaché au SID (la colonne SID de DAGSTAR) ou à assoc ?

# Jointure LIBELLE pour savoir à quel SID est rattaché chaque LID
# (un LID peut être attaché à plusieurs SID en théorie ; on prend tous les rattachements)
lid_to_sids = (
    ctx.ofs["libelle"]
    .group_by("LID")
    .agg(pl.col("SID").alias("sids_rattaches"))
)

dagstar_check = dagstar.join(lid_to_sids, on="LID", how="left")

# Pour chaque ligne, vérifier si le LID est rattaché au SID (de DAGSTAR) ou à assoc
def position_lid(row):
    if row["sids_rattaches"] is None:
        return "lid_introuvable"
    sids = row["sids_rattaches"]
    if row["SID"] in sids and row["assoc"] in sids:
        return "lid_aux_deux"
    elif row["SID"] in sids:
        return "lid_cote_SID"
    elif row["assoc"] in sids:
        return "lid_cote_assoc"
    else:
        return "lid_ailleurs"

dagstar_check = dagstar_check.with_columns(
    pl.struct(["SID", "assoc", "sids_rattaches"])
    .map_elements(position_lid, return_dtype=pl.String)
    .alias("position_lid")
)
distrib_pos = (
    dagstar_check
    .group_by("position_lid")
    .agg(pl.len().alias("n"))
    .sort("n", descending=True)
)
print("\nPosition du LID par rapport aux SID de la ligne DAGSTAR :")
print(distrib_pos)

# Croiser avec daget pour voir si la position dépend du niveau
print("\nCroisé position_lid × daget :")
distrib_pos_daget = (
    dagstar_check
    .group_by(["daget", "position_lid"])
    .agg(pl.len().alias("n"))
    .sort(["daget", "n"], descending=[False, True])
)
print(distrib_pos_daget)

# =====================================================================
# Synthèse
# =====================================================================
print("\n" + "=" * 70)
print("SYNTHÈSE")
print("=" * 70)
print(f"DAGSTAR : {total_dagstar} lignes, {total_lid_uniques} LID uniques")
print(f"LID couverts par LIBELLE/INCLUDE/EXCLUDE/DESCR : {len(couverts)} ({100*len(couverts)/len(lid_dagstar):.1f}%)")
print(f"LID non couverts : {len(non_couverts)}")
print(f"Distribution source LIBELLE : voir tableau ci-dessus")
print(f"Position du LID : voir tableau ci-dessus")
```

# INVESTIGATION — LID DAGSTAR avec source=S (libellé systématique)

```python
import polars as pl

# =====================================================================
# Investigation approfondie : les 625 LID DAGSTAR de source=S
# =====================================================================
print("=" * 70)
print("INVESTIGATION — LID DAGSTAR avec source=S (libellé systématique)")
print("=" * 70)

dagstar = ctx.ofs["dagstar"]
libelle = ctx.ofs["libelle"]
master = ctx.ofs["master"]

# Jointure DAGSTAR avec LIBELLE pour récupérer le source
dagstar_with_source = dagstar.join(
    libelle.select(["LID", "source", "libelle"]),
    on="LID",
    how="left",
)

# Filtrer les lignes où le LID a source=S
dag_s = dagstar_with_source.filter(pl.col("source") == "S")
print(f"\nNombre de lignes DAGSTAR avec LID source=S : {dag_s.height}")
print(f"Nombre de LID uniques source=S : {dag_s.select('LID').n_unique()}")

# Distribution par daget
print("\n--- Question 2 : corrélation avec daget ---")
print(dag_s.group_by("daget").agg(pl.len().alias("n")).sort("n", descending=True))

# Comparons aussi avec les LID source=D pour voir la différence
print("\nPour comparaison, distribution daget des LID source=D :")
dag_d = dagstar_with_source.filter(pl.col("source") == "D")
print(dag_d.group_by("daget").agg(pl.len().alias("n")).sort("n", descending=True))

# =====================================================================
# Question 3 — Le LID systématique est-il référencé plusieurs fois ?
# =====================================================================
print("\n--- Question 3 : nombre de paires DAGSTAR par LID source=S ---")
# Pour chaque LID source=S, combien de lignes DAGSTAR pointent vers lui ?
freq_lid_s = (
    dag_s
    .group_by("LID")
    .agg(pl.len().alias("nb_paires"))
    .sort("nb_paires", descending=True)
)
print(f"\nDistribution du nombre de paires par LID source=S :")
print(freq_lid_s.select(
    pl.col("nb_paires").min().alias("min"),
    pl.col("nb_paires").quantile(0.5).alias("median"),
    pl.col("nb_paires").mean().alias("mean"),
    pl.col("nb_paires").max().alias("max"),
))

print("\nTop 10 des LID source=S avec le plus de paires :")
top_lid_s = freq_lid_s.head(10).join(
    libelle.filter(pl.col("source") == "S").select(["LID", "SID", "libelle"]),
    on="LID",
    how="left",
).join(
    master.select(["SID", "code"]),
    on="SID",
    how="left",
)
print(top_lid_s)

# =====================================================================
# Question 1 + 4 — Types de codes concernés et exemples
# =====================================================================
print("\n--- Question 1 : types de codes concernés ---")
# Joindre avec master pour récupérer le code et le type
dag_s_enriched = dag_s.join(
    libelle.select(["LID", "SID"]).rename({"SID": "SID_libelle"}),
    on="LID",
    how="left",
).join(
    master.select(["SID", "code", "type", "level"]).rename({"SID": "SID_libelle"}),
    on="SID_libelle",
    how="left",
)

print("\nDistribution du type de code (master.type) pour les LID source=S de DAGSTAR :")
print(dag_s_enriched.group_by("type").agg(pl.len().alias("n")).sort("n", descending=True))

print("\nDistribution du level (master.level) pour les LID source=S de DAGSTAR :")
print(dag_s_enriched.group_by("level").agg(pl.len().alias("n")).sort("n", descending=True))

# Quelques exemples concrets variés
print("\n--- Question 4 : 10 exemples concrets ---")
echantillon = dag_s_enriched.sample(n=min(10, dag_s_enriched.height), seed=42).select([
    "SID", "LID", "assoc", "daget", "plus", "code", "level"
])
print(echantillon)

# Pour chaque exemple, montrer le libellé et chercher le code associé
print("\n--- Détail des 10 exemples (code principal, libellé, code associé) ---")
for row in echantillon.iter_rows(named=True):
    sid = row["SID"]
    lid = row["LID"]
    assoc = row["assoc"]
    daget = row["daget"]
    plus = row["plus"]
    code = row["code"]
    
    # Libellé du LID
    lib_row = libelle.filter(pl.col("LID") == lid).select("libelle").limit(1)
    libelle_text = lib_row.row(0)[0] if lib_row.height > 0 else "(introuvable)"
    
    # Code de l'assoc
    code_assoc_row = master.filter(pl.col("SID") == assoc).select(["code", "level"]).limit(1)
    code_assoc = code_assoc_row.row(0)[0] if code_assoc_row.height > 0 else f"SID={assoc}"
    
    print(f"\n  Code principal : {code} (SID={sid}, daget={daget}, plus={plus})")
    print(f"  LID {lid} : {libelle_text[:80]}")
    print(f"  Code associé   : {code_assoc} (SID={assoc})")
```
